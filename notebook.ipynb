```python
from pinecone import Pinecone, ServerlessSpec
import os
from dotenv import load_dotenv

load_dotenv()

api_key = os.getenv("PINECONE_API_KEY")

pc = Pinecone(
    api_key=api_key,
)

dense_index_name = "rag-actuaria-2"
if not pc.has_index(dense_index_name):
    pc.create_index(
        name=dense_index_name,
        dimension=1536,
        metric="cosine",
        spec=ServerlessSpec(
            cloud='aws',
            region='us-east-1'
        )
    )

dense_index = pc.Index(dense_index_name)
dense_index.describe_index_stats()
```

```python
from chonkie.pipeline import Pipeline
from chonkie import OpenAIEmbeddings
import os
from dotenv import load_dotenv

load_dotenv()

embeddings = OpenAIEmbeddings(api_key=os.getenv("OPENAI_API_KEY"))

# Fetch all files from both directories
dirs = [
    "./sample_docs/global_docs",
    "./sample_docs/PANDUAN LENGKAP IMBALAN KERJA PSAK 219"
    # Add more directories if needed
]
docs = []
for dir_path in dirs:
    docs.extend(
        Pipeline()
        .fetch_from("file", dir=dir_path, ext=[".md"])
        .process_with("markdown")
        .chunk_with("late", chunk_size=2048)
        .refine_with("overlap", context_size=100)
        .run()
    )

# print(f"Processed {len(docs)} documents")
# for doc in docs:
#     print(f"  - {len(doc.chunks)} chunks")
```
